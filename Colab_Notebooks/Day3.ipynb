```python
# Q5. Find the factorial of a number using a loop.

num = 5
fact = 1

if num < 0:
    print("Factorial is not defined for negative numbers")
else:
    for i in range(1, num + 1):
        fact = fact * i

    print("Factorial:", fact)

# Output: Factorial: 120
```

```python
# Q6. Count the number of digits in a number.

num = 12345
num = abs(num)
count = 0

if num == 0:
    count = 1
else:
    while num > 0:
        count += 1
        num //= 10

print("Number of digits:", count)

# Output: Number of digits: 5
```

```python
# Q7. Reverse a number using a loop.

num = 12345
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num //= 10

print("Reversed number:", reverse)

# Output: Reversed number: 54321
```

```python
# Q8. Check whether a number is prime.

num = 17
is_prime = True

if num < 2:
    is_prime = False
else:
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break

if is_prime:
    print("Prime number")
else:
    print("Not a prime number")

# Output: Prime number
```

```python
# Q9. Write a function to check whether a number is palindrome.

def is_palindrome(num):
    if num < 0:
        return False

    original = num
    reverse = 0

    while num > 0:
        digit = num % 10
        reverse = reverse * 10 + digit
        num //= 10

    return original == reverse

print(is_palindrome(121))

# Output: True
```

```python
# Q10. Write a function to calculate the sum of digits of a number.

def sum_of_digits(num):
    num = abs(num)
    total = 0

    while num > 0:
        total += num % 10
        num //= 10

    return total

print("Sum:", sum_of_digits(12345))

# Output: Sum: 15
```

```python
# Q14. Reverse a list without using built-in functions.

numbers = [1, 2, 3, 4, 5]
reverse_list = []

for i in range(len(numbers) - 1, -1, -1):
    reverse_list.append(numbers[i])

print(reverse_list)

# Output: [5, 4, 3, 2, 1]
```

```python
# Q27. Create a Rectangle class to calculate area and perimeter.

class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)


r1 = Rectangle(10, 5)

print("Area:", r1.area())
print("Perimeter:", r1.perimeter())

# Output:
# Area: 50
# Perimeter: 30
```

```python
# Q28. Create an Employee class with salary
# and a method to increase salary.

class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def increase_salary(self, percentage):
        self.salary += self.salary * percentage / 100

    def display(self):
        print("Name:", self.name)
        print("Salary:", self.salary)


e1 = Employee("Rahul", 30000)

e1.display()
e1.increase_salary(10)

print("After increment:")
e1.display()

# Output:
# Name: Rahul
# Salary: 30000
# After increment:
# Name: Rahul
# Salary: 33000.0
```

```python
# Q29. Create a ShoppingCart class using a dictionary
# (item: price) and calculate the total bill.

class ShoppingCart:
    def __init__(self):
        self.items = {}

    def add_item(self, item, price):
        self.items[item] = price

    def calculate_total(self):
        return sum(self.items.values())

    def display(self):
        print("Items:", self.items)
        print("Total bill:", self.calculate_total())


cart = ShoppingCart()

cart.add_item("Book", 250)
cart.add_item("Pen", 20)
cart.add_item("Bag", 800)

cart.display()

# Output:
# Items: {'Book': 250, 'Pen': 20, 'Bag': 800}
# Total bill: 1070
```

```python
# Q30. Create a Library class to store books in a list
# and add/remove books.

class Library:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)
        print(book, "added")

    def remove_book(self, book):
        if book in self.books:
            self.books.remove(book)
            print(book, "removed")
        else:
            print(book, "not found")

    def display_books(self):
        print("Books:", self.books)


library = Library()

library.add_book("Python")
library.add_book("Java")
library.add_book("C++")

library.display_books()

library.remove_book("Java")

library.display_books()

# Output:
# Python added
# Java added
# C++ added
# Books: ['Python', 'Java', 'C++']
# Java removed
# Books: ['Python', 'C++']
```
